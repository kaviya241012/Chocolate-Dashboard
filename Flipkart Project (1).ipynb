### **Chart Requirements**

#### **Average Rating per Brand**

```python
Average_Rating_per_Brand=df.groupby('Brand')['Rating'].sum().sort_values(ascending=False)
plt.figure(figsize=(10,6))
bars=plt.bar(Average_Rating_per_Brand.index, Average_Rating_per_Brand.values)
plt.xticks(rotation=45)
plt.xlabel('Brand')
plt.ylabel('Rating')
plt.show()
```

#### Samsung leads with a significant margin, suggesting a strong market presence and high purchase volume
Apple, realme, and OPPO follow, reflecting consistent user activity and feedback.
Brands like IQOO and Google Pixel show relatively fewer ratings, which may indicate niche targeting or lesser availability.

### **Price vs Rating Correlation chart**

```python
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(10,6))
sns.scatterplot(data=df, x='Selling Price', y='Rating', hue='Brand')
plt.xlabel('Selling Price (₹)')
plt.ylabel('Rating')
plt.grid(True)
plt.tight_layout()
plt.show()
```

#### This scatter plot compares the selling price and user ratings of mobile phones across popular brands. 
Each point represents a phone model, colored by its brand. Phones between ₹10K–₹30K dominate the dataset and hold strong ratings.
Premium phones (₹60K+) show consistently high ratings — value meets expectation. 
Mid-range players like realme, Xiaomi, vivo are tightly clustered with 4.2+ ratings — budget kings!
